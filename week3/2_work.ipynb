```python
daily_prices = raw.groupby(["item", "date"])["price"].mean().reset_index()

daily_prices
```

```python
daily_prices = daily_prices.sort_values("date")
```

```python
first_last = daily_prices.groupby("item")["price"].agg(
    first_price="first",
    last_price="last"
)

first_last
```

```python
first_last["percentage_change"] = (
    (first_last["last_price"] - first_last["first_price"])
    / first_last["first_price"]
) * 100

first_last
```

```python
summary = raw.groupby("item")["price"].agg(
    Mean="mean",
    Median="median"
).round(2)

summary["First-to-last change (%)"] = first_last["percentage_change"].round(2)

summary
```

```python
import pandas as pd

week2 = pd.read_csv(r"C:\Users\user\Downloads\teaching_prices_messy.csv")

week2.head()
```

```python
week2 = week2.drop_duplicates()

week2["item"] = week2["item"].str.strip().str.title()

week2["price"] = pd.to_numeric(
    week2["price"].astype(str).str.replace(",", ""),
    errors="coerce"
)
```

```python
staple = week2[week2["item"] == "Staple A"].copy()

staple
```

```python
mean_with = staple["price"].mean()

median_with = staple["price"].median()

print("Mean with flagged value:", round(mean_with, 2))
print("Median with flagged value:", round(median_with, 2))
```

```python
staple_without = staple[staple["price"] != 1250].copy()

mean_without = staple_without["price"].mean()

median_without = staple_without["price"].median()

print("Mean without flagged value:", round(mean_without, 2))
print("Median without flagged value:", round(median_without, 2))
```

```python
comparison = pd.DataFrame({
    "Condition": ["With flagged value", "Without flagged value"],
    "Mean": [mean_with, mean_without],
    "Median": [median_with, median_without]
})

comparison.round(2)
```
